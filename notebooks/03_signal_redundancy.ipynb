# Signed effective scores, final weights and Pareto priority

SYNTHETIC METHOD DEMONSTRATION. Adapted from reviewed R1/R2 methods. No historical market or platform evidence is embedded.

An exact duplicate and its opposite are different decisions. Compare full continuous histories.

```python
from alpha_discovery.synthetic import make_panel
from alpha_discovery.local_first_evaluator import DualEvaluator
from alpha_discovery.signal_first import extract_books, exact_daily, weight_distance, quality_order
import numpy as np
panel, *_ = make_panel()
evaluator = DualEvaluator(panel)
raw = panel.fields['volume']
ix = np.flatnonzero((panel.dates >= '2019-01-01') & (panel.dates <= '2023-12-31'))
a = extract_books(raw, ix, 1, panel, evaluator)
b = extract_books(raw, ix, -1, panel, evaluator)
j = np.flatnonzero(a['defined'])[0]
assert exact_daily(a['score'][j], a['score'][j])[0] > .999
assert exact_daily(a['score'][j], b['score'][j])[0] < -.999
assert weight_distance(a['weights'][j], b['weights'][j]) > .999
print({'mode': 'SYNTHETIC', 'opposite_signed_rho': exact_daily(a['score'][j], b['score'][j])[0]})
```
