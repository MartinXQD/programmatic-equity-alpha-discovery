# Historical membership and accounting availability

SYNTHETIC METHOD DEMONSTRATION. Adapted from reviewed R1/R2 methods. No historical market or platform evidence is embedded.

Generate wholly artificial securities and accounts. No market records are distributed.

```python
from alpha_discovery.synthetic import synthetic_inputs
from alpha_discovery.preprocessing import monthly_membership, apply_membership, attach_accounts
raw, accounts = synthetic_inputs()
membership = monthly_membership(raw, top_n=120)
selected = apply_membership(raw, membership)
assert (selected.formation_date < selected.date).all()
selected = attach_accounts(selected, accounts)
assert (selected.loc[selected.available_date.notna(), 'available_date'] <= selected.loc[selected.available_date.notna(), 'date']).all()
print({'mode': 'SYNTHETIC', 'selected_rows': len(selected), 'accounts': len(accounts)})
```

The 180-day proxy and stale report mask address different dates. Missing debt must stay missing.

```python
import numpy as np
assert selected.loc[(selected.date-selected.datadate).dt.days > 730, 'roa'].isna().all()
assert selected.loc[selected.dlc.isna(), 'leverage'].isna().all()
selected[['date','permno','formation_date','datadate','available_date','roa']].tail()
```
